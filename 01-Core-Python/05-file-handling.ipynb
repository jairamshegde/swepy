Python gives us several ways to format our output, each with different strengths:
1. simple string concatenation (manual)
2. f-strings (modern, preferred)
3. `.format()` (flexible way)

[NOTE] : personally I prefer f string in most of the cases!

```python
# lets try simple exercise
product = "laptop"
price = 799.99
quanity = 3
# print : you have purchased 3 laptop(s) for $799.99 each.

print(f"you have purchased {quantity} {product}(s) for ${price} each.")
```

### controlling the decimal values

```python
# What if we want to format the price to always show 2 decimal places?
price = 999.99
quantity = 3
total = price * quantity

# Try predicting what each of these will print:
print(f"Price: ${price:.2f}")          # The :.2f formats as float with 2 decimals
print(f"Total: ${total:.2f}")          # What if total was 2999.97?
print(f"Total: ${price * quantity:.2f}")  # We can even do math inside!

price = 1000
print(f"Price: ${price:.2f}")  # Will this show $1000 or $1000.00? --> 1000.00
```

### Debug Format

```python
# Traditional debugging
quantity = 3
product = "laptop"
price = 999.99

# The old way - repetitive
print(f"quantity: {quantity}, product: {product}, price: {price}")

# The new way - so much cleaner!
print(f"Debug: {quantity=} {product=} {price=}")
```

## str vs repr

```python
text = "Hello\nWorld"
print(str(text))      # Human-friendly
```

```python
print(repr(text))     # Shows exactly what the string contains
```

```python
# In f-strings, you can choose which one:
print(f"\nDefault (str): {text}")
print(f"With !r (repr): {text!r}") # focus on the expression
print(f"With !s (str): {text!s}")
```

### file path issue (windows)

```python
windows_path = "C:\\Users\\Alice\\Documents"
mixed_path = "C:\Users\Alice\Documents"
```

```python
print("What user sees (str):")
print(f"Path 1: {windows_path}")
print(f"Path 2: {mixed_path}")
```

```python
print("\nWhat debugger sees (repr):")
print(f"Path 1: {windows_path!r}")
print(f"Path 2: {mixed_path!r}")
```

### escaping sequences

```python
# The backslash (\) is special in Python strings
print("Common escape sequences:")
print("\\n = newline:", repr('\n'))
print("\\t = tab:", repr('\t'))
print("\\r = carriage return:", repr('\r'))
print("\\b = backspace:", repr('\b'))

# The troublemakers in Windows paths:
print("\nDangerous combinations in paths:")
print("\\U = Unicode (expects 8 digits):", "C:\\Users")  # \U starts Unicode
print("\\n = newline:", "C:\\new")                       # \n becomes newline
print("\\b = backspace:", "C:\\bin")                     # \b becomes backspace
```

[SOLUTION] to escape the escaping sequence use `\`.

## File Read - Write Ops

```python
# Writing to a file - the modern way
with open('test.txt', 'w', encoding='utf-8') as f:
    f.write("Hello, file world!\n")
    f.write(f"Written at {__import__('datetime').datetime.now()}\n")

print("File created! Check your directory.")
```

```python
# Reading the file
with open('test.txt', 'r', encoding='utf-8') as f:
    content = f.read()
    print(f"File contents:\n{content}")
```
