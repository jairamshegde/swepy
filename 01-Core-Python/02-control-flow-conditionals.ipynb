## For loop

1. The for statement in Python differs a bit from what you may be used to in C or other languages.
2. Python’s for statement iterates over the items of any sequence (a list or a string), in the order that they appear in the sequence.

````python
# measure each word length and print the length infornt
words = ['cat', 'window', 'defenestrate']
for word in words:
  print(word,len(word))
````

[WARNING]
Code that modifies a collection while iterating over that same collection can be tricky to get right.

````python
# Create a sample collection
users = {'Ram': 'active', 'Raj': 'inactive', 'Rani': 'active'}

# Strategy:  Iterate over a copy
for user, status in users.copy().items():
    if status == 'inactive':
        del users[user]
````

## Range

If you do need to iterate over a sequence of numbers `range(start,stop, step)` comes handy. step must be always an `integer`

````python
for i in range(5): #
  print(i)
````

### properties
1. by default we give the stop argument and stop never included the in the sequence : `[start,step)`
2.  It is possible to let the range start at another number, or to specify a different increment (even negative; sometimes this is called the ‘step’)

````python
list(range(5, 10))
````

````python
list(range(0, 10, 3))
````

````python
list(range(-10, -100, -30))
````

### NOTE

To iterate over the indices of a sequence, you can combine range() and len() as follows

````python
a = ['Mary', 'had', 'a', 'little', 'lamb']
for i in range(len(a)):
    print(i, a[i])
````

[TIP]
In most such cases, however, it is convenient to use the `enumerate()` function

````python
for idx, val in enumerate(a,start=0):
  print(idx,val)
````

### Caution

````python
print(range(10)) # range(0,10)
````

1. In many ways the object returned by `range()` behaves as if it is a list, but in fact it isn’t.
2. It is an object which returns the successive items of the desired sequence when you iterate over it, but **it doesn’t really make the list**, thus saving space.
3. We say such an object is `iterable`, that is, suitable as a target for functions and constructs that expect something from which they can obtain successive items until the supply is exhausted.

## Break and Continue Statements
1. The break statement breaks out of the innermost enclosing for or while loop.

````python
for n in range(2, 10):

    for x in range(2, n):

        if n % x == 0:

            print(n, 'equals', x, '*', n//x)

            break

    else:

        # loop fell through without finding a factor

        print(n, 'is a prime number')
````

## Match

Analogy:
1. For loops: Don't make me manage indices, just give me the items
2. Match statements: Don't make me extract pieces, let me describe the pattern  

````python
def handle_api_response(response):
    """Process different API response formats"""
    match response:
        case {"status": 200, "data": data}:
            return ("success", data)

        case {"status": 404}:
            return ("not_found", None)

        case {"status": 401 | 403}:  # Multiple values with |
            return ("unauthorized", None)

        case {"status": code, "error": {"message": msg}} if code >= 400:
            # Guard clause (if condition)
            return ("error", f"{code}: {msg}")

        case {"status": code} if code >= 500:
            return ("server_error", f"Server error: {code}")

        case _:
            return ("unknown", "Invalid response format")

# Test API responses
api_responses = [
    {"status": 200, "data": {"user": "john", "id": 123}},
    {"status": 404},
    {"status": 401},
    {"status": 400, "error": {"message": "Invalid input"}},
    {"status": 503},
    {"weird": "response"}
]

print("API Response Handling:")
for resp in api_responses:
    result = handle_api_response(resp)
    print(f"{resp} -> {result}")
````

### Rule of thumb
If you find yourself writing multiple `if` statements checking both the type/structure `AND` extracting values from the same data, that's when `match` shines.

```txt
WITHOUT MATCH, you see this pattern:
    if isinstance(data, dict):              # CHECK structure
        if 'key' in data:                   # CHECK presence
            value = data['key']             # EXTRACT
            if 'nested' in value:           # CHECK nested structure
                nested_val = value['nested'] # EXTRACT again
                # Finally do something with nested_val

WITH MATCH, it becomes:
    case {'key': {'nested': nested_val}}:   # CHECK + EXTRACT in one line
        # Immediately use nested_val

The more checking AND extracting you do, the more match shines!
```

````python
# Understanding how match handles dictionary patterns with extra keys

def process_user_data(data):
    """Demonstrate that match only checks for specified keys"""
    match data:
        case {"name": name, "age": age}:
            return f"User {name} is {age} years old"
        case {"name": name}:
            return f"User {name} (age unknown)"
        case _:
            return "Invalid user data"

# Test with different dictionaries
test_cases = [
    # Exact match - only required keys
    {"name": "Alice", "age": 30},

    # Extra keys - still matches!
    {"name": "Bob", "age": 25, "email": "bob@example.com", "city": "NYC"},

    # Many extra keys
    {"name": "Charlie", "age": 35, "email": "charlie@test.com",
     "phone": "123-456-7890", "address": "123 Main St", "country": "USA"},

    # Missing required key
    {"name": "David", "email": "david@example.com"},

    # No matching keys
    {"email": "only@email.com", "phone": "555-1234"}
]

print("Basic Dictionary Matching (extra keys are ignored):")
print("="*60)
for data in test_cases:
    result = process_user_data(data)
    print(f"Input: {data}")
    print(f"Result: {result}\n")
````
