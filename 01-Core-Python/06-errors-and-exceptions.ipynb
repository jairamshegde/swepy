# Python Exception Handling: From Print/None to Professional Error Management


## Your Learning Journey


### 1. **Where You Started**
You were using defensive programming:


````python
# Your original pattern
if condition:
    # do something
else:
    print("error message")
    return None
````

**Problem**: This hides WHERE and WHY things failed.


### 2. **The Aha! Moment - Scripts vs Modules**

```txt
Application Structure:
├── main.py          (SCRIPT - makes decisions)
├── api_handler.py   (MODULE - bubbles up errors)
├── data_processor.py (MODULE - bubbles up errors)
└── file_reader.py   (MODULE - bubbles up errors)
```

**Key Insight**: Modules don't know the business context - only the caller does!


### 3. **The Bubble Up Principle**
When an error happens deep in your code:
file_reader.py    → Error! → ↑
data_processor.py → (no catch) → ↑  
api_handler.py    → (no catch) → ↑
main.py           → catch & decide! ✓


**Why?** Because main.py knows:
- Is this error critical?
- Should we retry?
- Use a fallback?
- Notify the user?

1. **Let it bubble** (most common for modules)

2. **Add context, then bubble**


````python
except json.JSONDecodeError as e:
    raise ValueError(f"Config file has invalid JSON: {e}") from e
````

3. **Handle completely** (only if module can fix it)


````python
def read_with_fallback(primary, backup):
    try:
        return read_file(primary)
    except FileNotFoundError:
        return read_file(backup)  # Module has a solution!
````

### 6. **The Decision Framework**
Ask yourself:
- **Can this module completely fix the problem?**
  - YES → Handle it
  - NO → Bubble it up with context

- **Is this error expected in this module?**
  - YES → Handle it (like optional cache files)
  - NO → Bubble it up

- **Does the module have enough context to decide?**
  - YES → Make the decision
  - NO → Let the caller decide

### 8. **The Professional Mindset**

- **Errors are not failures** - they're communication

- **None is not an error** - it hides information
