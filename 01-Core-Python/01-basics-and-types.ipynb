### Indexing

`0 based indexing`
The index tells you how many steps to take from that starting point.

Strings can be indexed (subscripted), with the first character having index 0.


```python
word = 'Python'
print(word[0])
print(word[1])
```

negative indices are just shortcuts and it is calculated using `len(size of the sequence) -1` would shown as `[-1]`

```python
print(word[-1])
print(word[-6])
```

In addition to indexing, slicing is also supported. While indexing is used to obtain individual characters, slicing allows you to obtain a substring

### slicing

```python
print(word[0:2]) # characters from position 0 (included) to 2 (excluded)
print(word[2:5]) # characters from position 2 (included) to 5 (excluded)
```

```python
print(word[:2])   # character from the beginning to position 2 (excluded)
print(word[4:])  # characters from position 4 (included) to the end
```

```python
print(word[-2:]) # characters from the second-last (included) to the end
```

NOTE:
Note how the start is always included, and the end always excluded.

```python
word[:2] + word[2:]
```

One way to remember how slices work is to think of the indices as pointing between characters, with the left edge of the first character numbered 0. Then the right edge of the last character of a string of n characters has index n, for example:

### strings are Immutable

Python strings cannot be changed — they are immutable. Therefore, assigning to an indexed position in the string results in an error:

```python
word[0] = 'J'
```

If you need a different string, you should create a new one:

```python
'J' + word[1:]
```

Simple assignment in Python never copies data. When you assign a list to a variable, the variable refers to the existing list. Any changes you make to the list through one variable will be seen through all other variables that refer to it.

```python
id(rgb) == id(rgba)  # they reference the same object
```

## Firs step towards programming

Fibonacci series: the sum of two elements defines the next
$$ 0,1,1,2,3,5,8 ⋯ $$

```python
a, b = 0,1 # unpacking, tuple unpacking : order matters!

while a < 10 :

  a,b = b, a + b

print(a)

```

### Explanation

1. The first line contains a multiple assignment (a will take 0 and b will take 1)
2. he right-hand side are all evaluated first before any of the assignments take place.
  1. The right-hand side expressions are evaluated from the left to the right.
3. `while` loop executes as long as condition is `True`
  1. any non-zero integer value is true; zero is false.
  2.  The condition may also be a string or list value, in fact any sequence; anything with a non-zero length is true, empty sequences are false.
4.Standard comparison operartors are
  1. `<` (less than)
  2. `>` (greater than)
  3. `==` (equal to)
  4. `!=` (not equal to)
  5. `<=` (less than or equals to)
  6. `>=` (greater than or equals to)

5. The body of the loop is indented: indentation is Python’s way of grouping statements. in most of the languages we use `{ }` to group the statements
