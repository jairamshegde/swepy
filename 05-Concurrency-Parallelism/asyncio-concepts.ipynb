# Asyncio: Asynchronous Programming in Python

Asyncio is a library to write concurrent code using the `async/await` syntax. It is used as a foundation for multiple Python asynchronous frameworks that provide high-performance network and web-servers, database connection libraries, distributed task queues, etc.

## 1. Coroutines and `async/await`

A coroutine is a specialized version of a Python generator function. It is a function that can suspend its execution before reaching `return`, and it can indirectly pass control to another coroutine for some time.

```python
import asyncio
import time

async def say_hello():
    print("Hello")
    await asyncio.sleep(1) # Simulates an I/O operation
    print("World")

# To run a coroutine, we need an event loop. 
# In Jupyter, an event loop is already running, so we can just await it if we were in another async function.
# But for standalone scripts, we use asyncio.run(say_hello())

# In Jupyter notebook environment:
await say_hello()
```

## 2. Tasks and Concurrency

To run coroutines concurrently, we wrap them in Tasks.

```python
async def brew_coffee():
    print("Start brewing coffee...")
    await asyncio.sleep(2)
    print("Coffee is ready!")
    return "Coffee"

async def toast_bread():
    print("Start toasting bread...")
    await asyncio.sleep(1)
    print("Toast is ready!")
    return "Toast"

async def main_breakfast():
    start_time = time.time()
    
    # Schedule both calls concurrently
    task1 = asyncio.create_task(brew_coffee())
    task2 = asyncio.create_task(toast_bread())
    
    # Wait for them to finish
    coffee = await task1
    toast = await task2
    
    end_time = time.time()
    print(f"Finished breakfast with {coffee} and {toast} in {end_time - start_time:.2f} seconds")

await main_breakfast()
```

## 3. `asyncio.gather`

A convenient way to run multiple awaitables concurrently.

```python
async def fetch_data(id):
    print(f"Fetching data {id}...")
    await asyncio.sleep(1)
    print(f"Data {id} fetched")
    return {"id": id, "data": "sample"}

async def main_gather():
    # Run 3 fetches concurrently
    results = await asyncio.gather(
        fetch_data(1),
        fetch_data(2),
        fetch_data(3)
    )
    print(f"All results: {results}")

await main_gather()
```

## 4. Blocking vs Non-Blocking

It's crucial not to use blocking calls (like `time.sleep` or heavy CPU calculations) inside async functions, as they block the entire event loop.

```python
async def blocking_task():
    # This blocks the loop! Nothing else can run.
    print("Blocking start")
    time.sleep(2) 
    print("Blocking end")

async def non_blocking_task():
    # This yields control, allowing other tasks to run.
    print("Non-blocking start")
    await asyncio.sleep(2)
    print("Non-blocking end")
```
