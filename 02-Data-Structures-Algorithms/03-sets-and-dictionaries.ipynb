## Set

 A set is an unordered collection with no duplicate elements. Basic uses include membership testing and eliminating duplicate entries.

 supported numerical operations are
 1. union
 2. intersection
 3. difference
 4. symmetric difference


 ### Creation quirks
 curly braces `{}` represents the set. Quirk is while creating an empty set it si compulsory to use `set()` and `{}` not allowed  

````python
basket = {'apple', 'orange', 'apple', 'pear', 'orange', 'banana'}
basket # show that duplicates have been removed
````

````python
'orange' in basket  # fast membership testing
````

### numeric operations

````python
a = set('abracadabra')
b = set('alacazam')
````

````python
a # unique letters in a
````

````python
a - b # letters in a but not in b
````

````python
a | b # Uinon : letters in a or b or both
````

````python
a & b # intersection:  letters in both a and b
````

````python
a ^ b # letters in a or b but not both
````

## Dictionary

It is best to think of a dictionary as a set of **key: value pairs**, with the requirement **that the keys are unique** (within one dictionary).

Unlike sequences, which are indexed by a range of numbers, dictionaries are indexed by keys.

The main operations on a dictionary are **storing a value with some key and extracting the value given the key**.

1. A pair of braces creates an empty dictionary: {}
2. Keys must be immutable data type : numers, strings
3. Tuples can be used as keys if they contain only strings, numbers, or tuples;
4. if a tuple contains any mutable object either directly or indirectly, it cannot be used as a key.
5. If you store using a key that is already in use, the old value associated with that key is forgotten.
6. It is an error to extract a value using a non-existent key.
7. Performing `list(d)` on a dictionary returns a list of all the keys used in the dictionary, in insertion order.
8.  To check whether a single key is in the dictionary, use the `in` operator

````python
tel = {'jack': 4098, 'sape': 4139}
````

````python
tel['guido'] = 4127
````

````python
tel['jack'] # get the value by using key
````

````python
list(tel)
````

````python
sorted(tel) # returns list
````

## Looping Techniques

### iterate throguh dict

````python
knights = {'gallahad': 'the pure', 'robin': 'the brave'}
knights.items() # dict_items([('gallahad', 'the pure'), ('robin', 'the brave')])

# in the looping, key value unpacking takes place
for k, v in knights.items():
  print(k,v)
````

### enumerate
When looping through a sequence, the position index and corresponding value can be retrieved at the same time.

````python
for i, v in enumerate(['tic', 'tac', 'toe']):
  print(i,v)
````

### zip
To loop over **two or more sequences** at the same time, the entries can be paired with the zip().

````python
questions = ['name', 'quest', 'favorite color']

answers = ['lancelot', 'the holy grail', 'blue']

print(zip(questions, answers))

for q, a in zip(questions, answers):
    print(f"What is your {q}?  It is {a}")
````

## sorting

````python
basket = ['apple', 'orange', 'apple', 'pear', 'orange', 'banana']
sorted(basket) # returns new list unaltering the source list
````

````python
for i in sorted(basket):
  print(i)
````

## More on Conditionals

### Memebership operator
1. `in`
2. `not in`

are membership tests that determine whether a value is in (or not in) a container.

### compare objects

1. `is`
2. `is not`
compare whether two objects are really the same object.

### Comaprison can be chained

For example, `a < b == c` tests whether `a` is less than `b` and moreover `b` equals `c`.

[NOTE]
All comparison operators have the same priority, which is lower than that of all numerical(arithmetic) operators.
