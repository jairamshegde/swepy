## using list as stack

characterised by two kind of operation, push and pop. follows the principle of Last In First Out(LIFO).

```python
stack = [1,2,3]
stack.append(4) # push operation
stack.pop() # pop operation removes last item (returns none)
stack[-1] # peak (last item)
```

### using list as queue

It is also possible to use a list as a queue, where the first element added is the first element retrieved (“first-in, first-out”)(FIFO); however,  lists are not efficient for this purpose.

While appends and pops from the end of list are fast, **doing inserts or pops from the beginning of a list is slow** (because all of the other elements have to be shifted by one).

```python
from collections import deque
queue = deque(["Eric", "John", "Michael"])
queue.append("Terry")           # Terry arrives
queue.append("Graham")          # Graham arrives
queue.popleft()                 # The first to arrive now leaves

queue.popleft()                 # The second to arrive now leaves

queue.pop()                     # remove from the right side
queue.                          # Remaining queue in order of arrival
```

### List Comprehension

A list comprehension consists of brackets containing an expression followed by a `for` clause, then zero or more `for` or `if` clauses

1. List comprehensions provide a **concise way to create lists**.
2. Make new lists where each element is the result of some operations applied to each member of another sequence or iterable
3. To create a subsequence of those elements that satisfy a certain condition.



```python
squares = []
for x in range(10):
    squares.append(x**2)
```

```python
squares = [x**2 for x in range(10)]
```

```python
[(x, y) for x in [1,2,3] for y in [3,1,4] if x != y] # as list comprehension runs from left to right, in general case it is going from outer to inner steps
```

```python
combs = []
for x in [1,2,3]:
    for y in [3,1,4]:
        if x != y:
            combs.append((x, y))
```

NOTE: the tuple must be parenthesized, otherwise an error is raised

## Tuples and Sequences
we have seen the fundamental sequence types like list and strings and they have many things in common. There is another sequence type called tuple.

A tuple consists of a number of values separated by commas

```python
t = 1,"hi",True
type(t)
```

```python
t[0] # can be indexed
```

```python
u = t, (1,2,3) # can be nested
```

```python
t[0] = 2 # tuples are immutable, we cant re assign the values at any position
```

```python
v = ([1,2,3],[3,4]) # can contain the mutable objects means v[0][0] = 100 possible
```

[NOTE] output tuples are always enclosed in parentheses, so that nested tuples are interpreted correctly, though input may not have the paranthesis

### keep in mind
1. list and tuples may seem similar but used in the different scenarios
2. Tuple contains (most of the time) heterogeneous sequence of elements that are accessed via **unpacking** or indexing.
3. Lists are mutable and contains homogenous sequences most of the cases and accessed via iteration
