# Stack and Queue in Python

This notebook covers the implementation of Stack and Queue data structures in Python, referring to the official documentation.

## 1. Stack (LIFO - Last In, First Out)

A stack is a data structure where the last element added is the first one to be removed. Think of a stack of plates.

### Using `list` as a Stack

The Python official documentation recommends using the built-in `list` as a stack. It is efficient for this purpose.

- **Push**: Use `list.append(x)` to add an item to the top.
- **Pop**: Use `list.pop()` to remove an item from the top.

> **Note**: `list.append` and `list.pop` (from the end) are O(1) operations.

```python
stack = []

# Push elements
stack.append('a')
stack.append('b')
stack.append('c')

print(f"Stack after pushes: {stack}")

# Pop elements
print(f"Popped: {stack.pop()}")
print(f"Popped: {stack.pop()}")
print(f"Popped: {stack.pop()}")

print(f"Stack after pops: {stack}")
```

## 2. Queue (FIFO - First In, First Out)

A queue is a data structure where the first element added is the first one to be removed. Think of a line at a ticket counter.

### Why NOT use `list` as a Queue?

While you *can* use `list.insert(0, x)` or `list.pop(0)`, it is **inefficient**. Doing inserts or pops from the beginning of a list is O(n) because all other elements have to be shifted in memory.

### Using `collections.deque`

The official documentation recommends `collections.deque` (double-ended queue) for implementing queues. It is designed to have fast appends and pops from both ends.

- **Enqueue**: Use `deque.append(x)`
- **Dequeue**: Use `deque.popleft()`

> **Note**: `deque.popleft()` is O(1).

```python
from collections import deque

queue = deque(["Eric", "John", "Michael"])
print(f"Initial queue: {queue}")

# Enqueue (Arriving)
queue.append("Terry")
queue.append("Graham")
print(f"Queue after arrivals: {queue}")

# Dequeue (Leaving)
print(f"Left: {queue.popleft()}")
print(f"Left: {queue.popleft()}")

print(f"Queue after departures: {queue}")
```
