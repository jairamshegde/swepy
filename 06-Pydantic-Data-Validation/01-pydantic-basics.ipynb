# Pydantic Basics

Pydantic is the most widely used data validation library for Python. It uses Python type hints to validate data and manage settings.

## Why Pydantic?
- **Type Validation**: Enforces type hints at runtime.
- **Performance**: Core validation logic is written in Rust (in V2), making it extremely fast.
- **JSON Schema**: Automatically generates JSON schema from models.
- **Ecosystem**: Used by FastAPI, LangChain, and many other modern Python tools.

## 1. The `BaseModel`

The core of Pydantic is the `BaseModel`. You define your data structure as a class inheriting from `BaseModel`.

```python
from pydantic import BaseModel

class User(BaseModel):
    id: int
    name: str
    is_active: bool = True  # Default value
```

## 2. Creating Instances & Dictionary Unpacking

You can create instances by passing arguments. A common pattern is unpacking a dictionary using `**`.

```python
user_data = {
    "id": 1,
    "name": "Alice",
    "is_active": True
}

# Correct way to unpack dictionary
user = User(**user_data)
print(user)
```

## 3. Data Validation & Type Coercion

Pydantic tries to coerce types where possible (e.g., string "123" to int 123). If it fails, it raises a `ValidationError`.

```python
from pydantic import ValidationError

try:
    # This will fail because 'name' expects a string, not a list
    User(id=2, name=["Not", "a", "string"])
except ValidationError as e:
    print(e)
```

## 4. The `Field` Function

For more complex validation (like min/max length, ranges) or metadata, use `Field`.

```python
from pydantic import Field

class Product(BaseModel):
    name: str = Field(..., min_length=3, max_length=50, description="Name of the product")
    price: float = Field(..., gt=0, description="Price must be greater than 0")
    quantity: int = Field(1, ge=1, description="Quantity, default is 1")

try:
    p = Product(name="A", price=-10)
except ValidationError as e:
    print(e)
```

### Common Field Parameters
- `default`: Default value (first argument).
- `...` (Ellipsis): Indicates the field is required.
- `min_length` / `max_length`: For strings.
- `gt` (greater than), `lt` (less than), `ge` (greater equal), `le` (less equal): For numbers.
- `pattern`: Regex validation.
- `description`: For documentation/schema generation.

## 5. Optional Fields

Use `Optional` (or `str | None` in Python 3.10+) for fields that can be `None`.

```python
from typing import Optional

class Employee(BaseModel):
    name: str
    department: Optional[str] = None  # Defaults to None if not provided

emp = Employee(name="Bob")
print(emp)
print(emp.department)  # None
```
