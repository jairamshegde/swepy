# Python Generators: The Lazy Iterators

Generators are a simple and powerful tool for creating iterators. They allow you to write iterators using a function-like syntax, avoiding the boilerplate code of creating a class with `__iter__()` and `__next__()` methods.

## 1. The Big Picture: Iterable vs Iterator vs Generator

To understand generators, we must recall the hierarchy:

1.  **Iterable**: An object capable of returning its members one at a time (e.g., lists, tuples, strings). It has an `__iter__()` method.
2.  **Iterator**: An object representing a stream of data. It has `__iter__()` and `__next__()` methods. It remembers its state.
3.  **Generator**: A specific **type of iterator** that is defined using a function with `yield`. It automatically implements the iterator protocol.

> **Key Concept**: All generators are iterators, but not all iterators are generators.

```python
import collections.abc

def my_generator():
    yield 1

gen = my_generator()

# Verification
print(f"Is generator an Iterator? {isinstance(gen, collections.abc.Iterator)}")
print(f"Is generator an Iterable? {isinstance(gen, collections.abc.Iterable)}")

# Check for methods
print(f"Has __iter__? {hasattr(gen, '__iter__')}")
print(f"Has __next__? {hasattr(gen, '__next__')}")
```

## 2. How Generators Simplify Iterators

Let's compare creating a custom iterator class vs using a generator.

```python
# --- The Hard Way: Custom Iterator Class ---
class SquareIterator:
    def __init__(self, max_n):
        self.max_n = max_n
        self.n = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.n < self.max_n:
            result = self.n ** 2
            self.n += 1
            return result
        else:
            raise StopIteration

# --- The Easy Way: Generator Function ---
def square_generator(max_n):
    for n in range(max_n):
        yield n ** 2

# Usage is identical
print("Class Iterator:", list(SquareIterator(5)))
print("Generator:", list(square_generator(5)))
```

## 3. The `yield` Keyword

When a function uses `yield`, it pauses execution and saves its state (local variables, instruction pointer). When `next()` is called again, it resumes exactly where it left off.

```python
def simple_gen():
    print("Start")
    yield 1
    print("Resumed after 1")
    yield 2
    print("Resumed after 2")
    yield 3
    print("Done")

g = simple_gen()
print(next(g)) # Runs until first yield
print(next(g)) # Resumes and runs until second yield
print(next(g)) # Resumes and runs until third yield
```

## 4. Generator Expressions

Just like list comprehensions, but lazy. They don't compute values until you ask for them.

```python
import sys

# List Comprehension (Eager)
my_list = [x * 2 for x in range(10000)]
print(f"List size in bytes: {sys.getsizeof(my_list)}")

# Generator Expression (Lazy)
my_gen = (x * 2 for x in range(10000))
print(f"Generator size in bytes: {sys.getsizeof(my_gen)}")
```

## 5. `yield from` (Delegating to Sub-generators)

Allows a generator to delegate part of its operations to another generator. It's like flattening nested iterators.

```python
def sub_gen():
    yield 'A'
    yield 'B'

def main_gen():
    yield 1
    yield from sub_gen() # Same as: for x in sub_gen(): yield x
    yield 2

print(list(main_gen()))
```
